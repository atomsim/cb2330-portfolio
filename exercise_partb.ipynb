```python
import numpy as np
import math
from scipy.stats import gmean
from scipy.special import factorial
```

```python
seq = "GCGCGCATATATA"
gc_count = 0

for base in seq:
  if base in ['G', "C"]:
    gc_count += 1

gc_content = gc_count/len(seq)
print(gc_content)

gc_content_np = np.sum(np.char.count(seq, ["G",'C']))/len(seq)
print(gc_content_np)
```

```python
n = 5
k = 2
bc = 0

def get_factorial(n):
  factorial = 1

  for i in range(1,n):
    factorial *= i
  return factorial

if k <= n:
  bc = get_factorial(n)/((get_factorial(k)) * get_factorial(n-k))

print(bc)
```
